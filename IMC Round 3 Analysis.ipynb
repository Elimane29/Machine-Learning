## VISUALISATION PRIX SPOT + PRIX OPTIONS (VOUCHERS)

```python
from bokeh.plotting import figure, show
from bokeh.models import ColumnDataSource, HoverTool
from bokeh.palettes import Category10
from bokeh.layouts import layout 


#Application de la transformée de fourier pour enlever fréquences parasites et lisser les courbes 

def filtre_fourier(signal, seuil= 0.01):
    fft_vals = np.fft.fft(signal)
    fft_freqs = np.fft.fftfreq(len(signal))
    fft_vals[np.abs(fft_freqs) > seuil] = 0
    signal_filtré = np.fft.ifft(fft_vals)
    return np.real(signal_filtré)

#Liste des produits

produits = ['VOLCANIC_ROCK' , 'VOLCANIC_ROCK_VOUCHER_9500', 'VOLCANIC_ROCK_VOUCHER_10000', 'VOLCANIC_ROCK_VOUCHER_10250','VOLCANIC_ROCK_VOUCHER_10500']


if 'symbol' not in df_vr.columns:
    raise KeyError("La colonne 'symbole' est absente du DataFrame")
#Initialisation Figure
p = figure(title = "Représentation Prix Spot et Prix Strike + Premium avec Lissage FFT et rapprochement spot des options pour observation", x_axis_label= "Timestamp", y_axis_label= "Mid Price",
           width = 1000, height = 500)

#Couleurs des courbes
colors = ['green', 'blue', 'orange', 'purple', 'red', 'black']

#Traçage de chaque produit , prod étant un produit dans la liste "produits"
prod = None
for i, prod in enumerate(produits):
   
    df_filtered = df_vr[df_vr['symbol'] == prod].copy()
    df_filtered['price']= filtre_fourier (df_filtered['price'])
    
    if prod == 'VOLCANIC_ROCK':
        df_filtered['price'] = df_filtered['price']
    elif prod == 'VOLCANIC_ROCK_VOUCHER_9500':
        df_filtered['price'] = df_filtered['price'] + 9500
    elif prod == 'VOLCANIC_ROCK_VOUCHER_9750':
        df_filtered['price'] = df_filtered['price'] + 9750
    elif prod == 'VOLCANIC_ROCK_VOUCHER_10000':
        df_filtered['price'] = df_filtered['price'] + 10000

    
    

    #Comme sur excel quand tu mets les données avec lesquelles tu vas faire ta figure 
    source = ColumnDataSource(df_filtered)

    line = p.line('timestamp' , 'price', source=source, line_width = 2, legend_label=prod, color = colors[i])

    hover  = HoverTool(tooltips = [ ("Produit" , prod) , ("Time", "@timestamp") , ("Mid Price", "@price{0.00}") ],
    renderers = [line])
    p.add_tools(hover)
p.legend.location = "top_left"
p.legend.click_policy = "hide" #Pour cacher/afficher les courbes
p.toolbar.autohide= True

show(p)
```
